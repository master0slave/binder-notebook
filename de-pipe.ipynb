```python
# Create New Data list
ชื่อ = ['เอ','บี','ซี','ดี','อี']
ขนาดเสื้อผ้า = ['M','L','S','XL','S']
ช่วงวัย = ['เด็ก','ผู้ใหญ่','วัยรุ่น','สูงอายุ','วัยรุ่น']
ระดับความรุนแรงของอาการ = ['อาการรุนแรง','อาการรุนแรง','ไม่มีอาการ','อาการเล็กน้อย','อาการปานกลาง']
ระดับการศึกษา = ['ประถมศึกษา','ปริญญาตรี','มัธยมศึกษา','ปริญญาเอก','ปริญญาตรี']

# Create DataFrame
df1 = pd.DataFrame({'ชื่อ':ชื่อ,'ขนาดเสื้อผ้า':ขนาดเสื้อผ้า,'ช่วงวัย':ช่วงวัย,'ระดับความรุนแรงของอาการ':ระดับความรุนแรงของอาการ,'ระดับการศึกษา':ระดับการศึกษา})
df1
```

```python
# Define order of column in DataFrame
order1 = ["S","M","L","XL"]
order2 = ["เด็ก","วัยรุ่น","ผู้ใหญ่","สูงอายุ"]
order3 = ["ไม่มีอาการ","อาการเล็กน้อย","อาการปานกลาง","อาการรุนแรง"]
order4 = ["ประถมศึกษา","มัธยมศึกษา","ปริญญาตรี","ปริญญาเอก"]
df1['ขนาดเสื้อผ้า'] = pd.Categorical(df1['ขนาดเสื้อผ้า'], categories=order1, ordered=True)
df1['ช่วงวัย'] = pd.Categorical(df1['ช่วงวัย'], categories=order2, ordered=True)
df1['ระดับความรุนแรงของอาการ'] = pd.Categorical(df1['ระดับความรุนแรงของอาการ'], categories=order3, ordered=True)
df1['ระดับการศึกษา'] = pd.Categorical(df1['ระดับการศึกษา'], categories=order4, ordered=True)
df1
```

```python
pd.get_dummies(df1, columns=['ขนาดเสื้อผ้า','ช่วงวัย','ระดับความรุนแรงของอาการ','ระดับการศึกษา'],dtype=int)
```

```python
# Exercise filling missing values
# Create list
from pandas import NA

รหัส = ['A001','B001','A005','C007','A003','B002']
เพศ = ['ชาย','ชาย','หญิง',NA,'ชาย', NA]
ส่วนสูง = [150,NA,165,160,155,162]
เกรด = [3.2,2.5,NA,3.9,NA,2.8]
ว่ายน้ำได้หรือไม่ = [NA,'ได้','ไม่ได้',NA,'ไม่ได้','ไม่ได้']

# Create DataFrame
df2 = pd.DataFrame({'รหัส':รหัส,'เพศ':เพศ,'ส่วนสูง':ส่วนสูง,'เกรด':เกรด,'ว่ายน้ำได้หรือไม่':ว่ายน้ำได้หรือไม่})
df2
```

```python
# Fill missing values with mean
df_imputed = df2.copy()
df_imputed['ส่วนสูง'].fillna(df_imputed['ส่วนสูง'].mean(), inplace=True)
df_imputed['เกรด'].fillna(df_imputed['เกรด'].mean(), inplace=True)

# Fill missing values with mode
df_imputed['เพศ'].fillna(df_imputed['เพศ'].mode()[0], inplace=True)
df_imputed['ว่ายน้ำได้หรือไม่'].fillna(df_imputed['ว่ายน้ำได้หรือไม่'].mode()[0], inplace=True)
df_imputed
```

```python
# Convert categorical data to numerical data
df_imputed['เพศ'] = pd.Categorical(df_imputed['เพศ'], categories=['หญิง','ชาย'], ordered=True)
df_imputed['เพศ'] = df_imputed['เพศ'].cat.codes
df_imputed['ว่ายน้ำได้หรือไม่'] = pd.Categorical(df_imputed['ว่ายน้ำได้หรือไม่'], categories=['ไม่ได้','ได้'], ordered=True)
df_imputed['ว่ายน้ำได้หรือไม่'] = df_imputed['ว่ายน้ำได้หรือไม่'].cat.codes
df_imputed
```

```python
# Normalize data
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df_normalized = df_imputed.copy()
df_normalized[['ส่วนสูง']] = scaler.fit_transform(df_normalized[['ส่วนสูง']])
df_normalized[['เกรด']] = df_normalized[['เกรด']]/4
df_normalized
```

```python
# Exercise 2
# Create list
รหัส = ['A001','B001','A005','C007','A003','B002']
เพศ = ['ชาย','ชาย','หญิง',NA,'ชาย', NA]
ส่วนสูง = [150,NA,165,160,155,162]
หมู่เลือด = ['AB','B','O',NA,'O','O']
เกรด = [3.2,2.5,NA,3.9,NA,2.8]
แพ้อาหารหรือไม่ = [NA,'แพ้','ไม่แพ้',NA,'ไม่แพ้','ไม่แพ้']

# Create DataFrame
df3 = pd.DataFrame({'รหัส':รหัส,'เพศ':เพศ,'ส่วนสูง':ส่วนสูง,'หมู่เลือด':หมู่เลือด,'เกรด':เกรด,'แพ้อาหารหรือไม่':แพ้อาหารหรือไม่})
df3
```

```python
# Fill missing values 
df3_imputed = df3.copy()
df3_imputed['เพศ'].fillna(df3_imputed['เพศ'].mode()[0], inplace=True)
df3_imputed['ส่วนสูง'].fillna(df3_imputed['ส่วนสูง'].mean(), inplace=True)
df3_imputed['หมู่เลือด'].fillna(df3_imputed['หมู่เลือด'].mode()[0], inplace=True)
df3_imputed['เกรด'].fillna(df3_imputed['เกรด'].mean(), inplace=True)
df3_imputed['แพ้อาหารหรือไม่'].fillna(df3_imputed['แพ้อาหารหรือไม่'].mode()[0], inplace=True)
df3_imputed
```

```python
# Convert categorical data to numerical data
df3_imputed= pd.get_dummies(df3_imputed, columns=['เพศ','หมู่เลือด','แพ้อาหารหรือไม่'],dtype=int)
df3_imputed
```

```python
# Normalize data
scaler = MinMaxScaler()
df3_normalized = df3_imputed.copy()
df3_normalized[['ส่วนสูง']] = scaler.fit_transform(df3_normalized[['ส่วนสูง']])
df3_normalized[['เกรด']] = df3_normalized[['เกรด']]/4
df3_normalized
```
